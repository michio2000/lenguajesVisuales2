## sentencia while(mientras)

se basa en repetir un bloque a partir de evaluar una condicion logica,
siempre que esta sea true 
queda en las manos del programador decidir el momento en que la condicion cambie a false para hacer que el while finalice


```python
c=0
while c<=5:
    c+=1
    print("c vale", c)
    
```

## instruccion break 
sirve para romper la ejecucuion del while en cualquier momento, no se ejecutara else, ya que solo se llama al finalizar la iteracion 


```python
c=0
while c<=5:
    c+=1
    if(c==4):
        print("rompemos el bucle cuando c vale",c)
        break
    print("c vale",c)
else:
    print("se ha completado toda la interaccion y c vale",c)
```

## instruccion continue
sirve para saltarse la interacion sin romper el bucle 

```python
c=0
while c<=5:
    c+=1
    if c==3 or c==4:
        #print("continuamos con la interacion ",c)
        continue 
    print("c vale",c)
else:
    print("se ha completado toda la interacion y c vale",c)
    
```

## creando un menu interactivo

```python
print("bienvenido al menu interactivo")
while(True):
    print("""que quiere hacer ? escribe una opcion
            1)Saludar
            2)sumar dos numeros
            3)salir """)
    opcion = input()
    if opcion =="1":
        print("Hola, espero estes bien !")
    elif opcion=="2":
        n1=float(input("ingrese el primer numero"))
        n2=float(input("ingrese el primer numero"))
        print("el resultado de la suma es ",n1+n2)
    elif opcion == '3':
        print("hasta luego ")
        break
    else:
        print("comando desconocido, vuelva a intentar")
```

que son las GUI y Tkinter?
- una GUI (Graphizal Use interface) es una interfaz que permite interactuar con un programa mediante ventanas, botones, campos, menus y otros elementos visuales.
- Tkinter es la biblioteca grafica estandar de python. normalmente ya viene incluida con python y  funciona sin conexion a internet

### la primera ventana : tk() y mainloop()
- import tkinter as tk : importarla biblioteca 
- tk.Tk() : Crea la ventana principal.
- title(): establece el titulo
- geometry(" ancho x alto ") = establece el tamaño inicial.
- configur(): modifica propiedades, como el color de fondo.
- mainloop(): mantiene la aplicacion esperando eventos.


## Ejemplo 1: Ventana Basica 

```python
import tkinter as tk 

ventana = tk.Tk()
ventana.title("Mi primera App")
ventana.geometry("500x300")
ventana.configure(bg="#EAF2F8")

#Mantiene abierta la ventana. Ejecuta una GUI por vez
ventana.mainloop()
```

## 3 Widests: los elementos  de la interfaz
Un **widgest es un elemento grafico. Los mas utilizados al comenzar son: 
- Label: Mostrar texto o informaciòn. 
- Button: Ejecutar una acciòn.
- Entry: Escribir una linea de texto.
- Text: Escribir Varias lineas de texto
- Checkbutton: Activar o desactivar una opciòn 
- Radiobutton: Elegir una opcion entre varias.
- Frame: Agrupar y desagrupar widgets

### los widgest reciben como primer argumento el contenedor al que pertenecen 

### Ejemplo 2: Etiqueta y boton

```python
import tkinter as tk 
def cambiar_mensaje():
    etiqueta.config(text="El botòn funciona", fg="#117864")

ventana = tk.Tk()
ventana.title("Etiqueta y boton")
ventana.geometry("400x220")

etiqueta = tk.Label(
    ventana,
    font=("Times New Roman",16),text="Presiona el boton ",
    
    fg="#1B4F72"
)
etiqueta.pack(pady=30)

boton = tk.Button(ventana, text="Cambiar mensaje", command=cambiar_mensaje)
boton.pack()

ventana.mainloop()
```

## 4. Gestores de geometria: pack(),grid() y place()
los widgest deben posicionarse dentro de la ventana.
- pack(): Diseños Sencillos, uno debajo de otro o por bloques
- grid(): Formularios organizados en filas y columnas.
- place(): Posiciones exactas con coordenadas; es menos adaptable 

Obs: Reglas importantes a tener en cuenta.
- No mezclar pack() y grid() dentro del **mismo contenedor**
- Si pueden utilizarse gestores diferentes en contenedores distintos.
- Para formularios, normalmente grid(), es la opcion mas clara.


```python
# Ejemplo 3: Formulario Organizado con grid()
import tkinter as tk
ventana = tk.Tk()
ventana.title("Formulario con grid")
ventana.geometry("450x230")

tk.Label(ventana, text="Nombre").grid(row=0,column=0,padx=10,pady=10,sticky="w")
entrada_nombre = tk.Entry(ventana,width=30)
entrada_nombre.grid(row=0, column=1, padx=25, pady=10)

tk.Label(ventana, text="Curso:").grid(row=1, column=0, padx=10, pady=10, sticky="e")
entrada_curso = tk.Entry(ventana,width=30)
entrada_curso.grid(row=1,column=1,padx=10,pady=10)

tk.Button(ventana, text="Guardar").grid(row=2, column=0, columnspan=2,pady=20)

ventana.mainloop()
```

## 5. Entrada y salida de datos con Entry 
- get() obtiene el contenido.
- delete(0, tk.END) borra el contenido.
- insert(posicion, texto) inserta texto.
- focus() coloca el cursor en el campo.

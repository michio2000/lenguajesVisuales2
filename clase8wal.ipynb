## 1. ¿Qué son una GUI y Tkinter?

Una **GUI** (*Graphical User Interface*) es una interfaz que permite interactuar con un programa mediante ventanas, botones, campos, menús y otros elementos visuales.

**Tkinter** es la biblioteca gráfica estándar de Python. Normalmente ya viene incluida con Python y funciona sin conexión a Internet.

### Comprobación

En una terminal puede ejecutarse:

```bash
python3 -m tkinter
```

Si aparece una ventana de prueba, Tkinter está disponible. En algunas distribuciones Linux puede ser necesario instalar el paquete `python3-tk`.

## 2. La primera ventana: `Tk()` y `mainloop()`

Toda aplicación comienza con una ventana principal.

- `import tkinter as tk`: importa la biblioteca.
- `tk.Tk()`: crea la ventana principal.
- `title()`: establece el título.
- `geometry("ancho x alto")`: establece el tamaño inicial.
- `configure()`: modifica propiedades, como el color de fondo.
- `mainloop()`: mantiene la aplicación esperando eventos.

### Ejemplo 1: ventana básica

````python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Mi primera ventana")
ventana.geometry("500x300")
ventana.configure(bg="#EAF2F8")

# Mantiene abierta la ventana. Ejecuta una GUI por vez.
ventana.mainloop()
````

## 3. Widgets: los elementos de la interfaz

Un **widget** es un elemento gráfico. Los más utilizados al comenzar son:

| Widget | Función |
|---|---|
| `Label` | Mostrar texto o información |
| `Button` | Ejecutar una acción |
| `Entry` | Escribir una línea de texto |
| `Text` | Escribir varias líneas |
| `Checkbutton` | Activar o desactivar una opción |
| `Radiobutton` | Elegir una opción entre varias |
| `Frame` | Agrupar y organizar widgets |

Los widgets reciben como primer argumento el contenedor al que pertenecen.

### Ejemplo 2: etiqueta y botón

````python
import tkinter as tk

def cambiar_mensaje():
    etiqueta.config(text="¡El botón funciona!", fg="#117864")

ventana = tk.Tk()
ventana.title("Etiqueta y botón")
ventana.geometry("400x220")

etiqueta = tk.Label(
    ventana,
    text="Presiona el botón",
    font=("Arial", 16),
    fg="#1B4F72"
)
etiqueta.pack(pady=30)

boton = tk.Button(ventana, text="Cambiar mensaje", command=cambiar_mensaje)
boton.pack()

ventana.mainloop()
````

### Conceptos importantes del ejemplo

- `text`: texto visible.
- `font`: tipo y tamaño de letra.
- `fg`: color del texto.
- `bg`: color de fondo.
- `config()`: modifica un widget después de crearlo.
- `command`: recibe la función que se ejecutará al hacer clic.

En `command=cambiar_mensaje` la función se escribe **sin paréntesis**, porque debe ejecutarse al producirse el clic y no al crear el botón.

## 4. Gestores de geometría: `pack()`, `grid()` y `place()`

Los widgets deben posicionarse dentro de la ventana.

| Gestor | Uso recomendado |
|---|---|
| `pack()` | Diseños sencillos, uno debajo de otro o por bloques |
| `grid()` | Formularios organizados en filas y columnas |
| `place()` | Posiciones exactas con coordenadas; es menos adaptable |

### Reglas importantes

- No mezclar `pack()` y `grid()` dentro del **mismo contenedor**.
- Sí pueden utilizarse gestores diferentes en contenedores distintos.
- Para formularios, normalmente `grid()` es la opción más clara.

### Ejemplo 3: formulario organizado con `grid()`

````python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Formulario con grid")
ventana.geometry("450x230")

tk.Label(ventana, text="Nombre:").grid(row=0, column=0, padx=10, pady=10, sticky="e")
entrada_nombre = tk.Entry(ventana, width=30)
entrada_nombre.grid(row=0, column=1, padx=10, pady=10)

tk.Label(ventana, text="Curso:").grid(row=1, column=0, padx=10, pady=10, sticky="e")
entrada_curso = tk.Entry(ventana, width=30)
entrada_curso.grid(row=1, column=1, padx=10, pady=10)

tk.Button(ventana, text="Guardar").grid(row=2, column=0, columnspan=2, pady=20)

ventana.mainloop()
````

`padx` y `pady` agregan espacio; `sticky` alinea el widget; `columnspan` permite que ocupe varias columnas.

## 5. Entrada y salida de datos con `Entry`

- `get()` obtiene el contenido.
- `delete(0, tk.END)` borra el contenido.
- `insert(posición, texto)` inserta texto.
- `focus()` coloca el cursor en el campo.

### Ejemplo 4: saludo personalizado

````python
import tkinter as tk

def saludar():
    nombre = entrada_nombre.get().strip()
    if nombre == "":
        resultado.config(text="Escribe tu nombre.", fg="red")
    else:
        resultado.config(text=f"¡Bienvenido/a, {nombre}!", fg="green")
        entrada_nombre.delete(0, tk.END)
        entrada_nombre.focus()

ventana = tk.Tk()
ventana.title("Saludo")
ventana.geometry("400x240")

tk.Label(ventana, text="Nombre:").pack(pady=(20, 5))
entrada_nombre = tk.Entry(ventana, width=30)
entrada_nombre.pack()
entrada_nombre.focus()

tk.Button(ventana, text="Saludar", command=saludar).pack(pady=15)
resultado = tk.Label(ventana, text="")
resultado.pack()

ventana.mainloop()
````

## 6. Variables de control

Tkinter posee variables que pueden conectarse con los widgets:

- `StringVar()` para texto.
- `IntVar()` para enteros.
- `DoubleVar()` para decimales.
- `BooleanVar()` para verdadero/falso.

Los métodos más usados son `get()` y `set()`.

### Ejemplo 5: elegir turno y aceptar una condición

````python
import tkinter as tk

def mostrar_seleccion():
    mensaje = f"Turno: {turno.get()} | Acepta condiciones: {acepta.get()}"
    resultado.config(text=mensaje)

ventana = tk.Tk()
ventana.title("Variables de control")
ventana.geometry("420x300")

turno = tk.StringVar(value="Mañana")
acepta = tk.BooleanVar(value=False)

tk.Label(ventana, text="Seleccione el turno:").pack(pady=10)
tk.Radiobutton(ventana, text="Mañana", variable=turno, value="Mañana").pack()
tk.Radiobutton(ventana, text="Noche", variable=turno, value="Noche").pack()
tk.Checkbutton(ventana, text="Acepto las condiciones", variable=acepta).pack(pady=10)
tk.Button(ventana, text="Mostrar", command=mostrar_seleccion).pack()
resultado = tk.Label(ventana, text="")
resultado.pack(pady=15)

ventana.mainloop()
````

## 7. Eventos: `command` y `bind()`

Un **evento** es una acción detectada por la aplicación.

- `command` se usa principalmente con botones.
- `bind()` permite responder al teclado, ratón u otros eventos.

Eventos frecuentes:

| Evento | Significado |
|---|---|
| `<Return>` | Tecla Enter |
| `<Escape>` | Tecla Escape |
| `<Button-1>` | Clic izquierdo |
| `<KeyRelease>` | Liberar una tecla |

La función asociada mediante `bind()` debe recibir el parámetro `evento`.

### Ejemplo 6: presionar Enter para saludar

````python
import tkinter as tk

def saludar(evento=None):
    nombre = entrada.get().strip()
    resultado.config(text=f"Hola, {nombre}" if nombre else "Falta el nombre")

ventana = tk.Tk()
ventana.title("Eventos")
ventana.geometry("380x200")

tk.Label(ventana, text="Escribe tu nombre y presiona Enter:").pack(pady=15)
entrada = tk.Entry(ventana, width=30)
entrada.pack()
entrada.bind("<Return>", saludar)
entrada.focus()

resultado = tk.Label(ventana, text="")
resultado.pack(pady=20)

ventana.mainloop()
````

## 8. Cuadros de diálogo con `messagebox`

Los cuadros de diálogo permiten comunicar información importante.

- `showinfo()`: información.
- `showwarning()`: advertencia.
- `showerror()`: error.
- `askyesno()`: pregunta que devuelve `True` o `False`.

### Ejemplo 7: validar y confirmar

````python
import tkinter as tk
from tkinter import messagebox

def guardar():
    nombre = entrada.get().strip()
    if nombre == "":
        messagebox.showwarning("Dato obligatorio", "Ingrese el nombre.")
        return
    messagebox.showinfo("Registro", f"{nombre} fue registrado correctamente.")

def cerrar():
    if messagebox.askyesno("Salir", "¿Desea cerrar la aplicación?"):
        ventana.destroy()

ventana = tk.Tk()
ventana.title("Mensajes")
ventana.geometry("380x210")

tk.Label(ventana, text="Nombre:").pack(pady=(20, 5))
entrada = tk.Entry(ventana, width=30)
entrada.pack()
tk.Button(ventana, text="Guardar", command=guardar).pack(pady=10)
tk.Button(ventana, text="Salir", command=cerrar).pack()

ventana.protocol("WM_DELETE_WINDOW", cerrar)
ventana.mainloop()
````

## 9. Organización con `Frame`

Un `Frame` es un contenedor utilizado para agrupar widgets. Ayuda a dividir la ventana en secciones y facilita combinar gestores de geometría sin mezclarlos en un mismo contenedor.

### Ejemplo 8: ventana dividida en secciones

````python
import tkinter as tk

ventana = tk.Tk()
ventana.title("Uso de Frame")
ventana.geometry("500x260")

encabezado = tk.Frame(ventana, bg="#1F618D")
encabezado.pack(fill="x")
tk.Label(encabezado, text="REGISTRO DE VISITANTES", bg="#1F618D", fg="white",
         font=("Arial", 15, "bold")).pack(pady=15)

formulario = tk.Frame(ventana, padx=20, pady=15, bg="yellow")
formulario.pack(expand=True, fill="y")

tk.Label(formulario, text="Nombre:").grid(row=0, column=0, sticky="e", pady=5)
tk.Entry(formulario, width=35).grid(row=0, column=1, padx=10, pady=5)
tk.Label(formulario, text="Documento:").grid(row=1, column=0, sticky="e", pady=5)
tk.Entry(formulario, width=35).grid(row=1, column=1, padx=10, pady=5)

botones = tk.Frame(ventana, pady=10, bg="red")
botones.pack()
tk.Button(botones, text="Guardar").pack(side="left", padx=5)
tk.Button(botones, text="Salir", command=ventana.destroy).pack(side="left", padx=5)

ventana.mainloop()
````

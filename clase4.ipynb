```python
cant=int(input("ingrese la cantidad de numeros que desee introducir"))
c=0
suma=0
while c<cant :
    n = int(input("ingrese el numero {}: ".format(c+1)))
    suma=suma+n
    c=c+1
else: 
    print("fin de la iteracion")
    promedio=suma/cant
print(" el promedio es ",promedio)
```

## retorno de valores 
para comunicarse con el exterior, las funciones pueden devolver valores al proceso principal 
gracias a la intruccion **return**
En el momento de devover un valor , la ejecuicion de la funcion finalizara

```python
def test():
    return " una cadena retornada "
test() 
```

## los valores devueltos se tratran como valores literales 
directos del tipo de dato retornado 

```python
print(test())
```

```python
c=test()
print(c)
```

```python
print(c+"10")
```

```python
def test():
    return(1,2,3,4,5)
test()
```

```python
print(test()[-1])
```

```python
print(test()[1:4])
```

```python
l=test()
```

```python
l[-1]
```

## parametros y argumentos 
EN La definicion de una funcion, los valores que se reciben se denominan parametros ,en la llamada se denominan argumentos 

# paso por valor y paso por referencia 
-** paso por valor **: se crea una copia de la variable dentro de una funcion.
-** paso por referencia **: se maneja directamente la variable. los cambios realizados 
dentro le afectaran tambien afuera

- conceptos claves de POO EN PYTHON
- Clase: Una planilla o molde que define las caracteristicas y acciones comunes que tendran los objetos 
- Objetos: Una instancia concreta de una clase que representa una entidad del mundo real (por ejemplo, un perro, un vehiculo)
- Atributo: Las variables que almacenan el estado o las propiedades (como el color o la edad)
- Metodos: las funciones definidas dentro de una clase que representan las acciones que el objeto puede realizar.

los cuatro pilares de la POO
- Encapsulamiento: Ocultar los datos internos de un objeto y protegerlos, permitiendo modificarlos solo a traves de metodos especificos 
- Herencia: Permite que una clase nueva herede atributos y metodos de otra clase existente para evitar repetir codigo.
- Polimorfismo: Usar un mismo metodo de distintas formas segun el objeto que lo este utilizando.
- Abstraccion: Simplicar conceptos complejos enfocandose solo en lo importante y ocultando los detalles tecnicos complejos

```python
class galleta():
    chocolate = False

    def __init__(self, sabor, forma):
        self.sabor = sabor
        self.forma = forma
        print("se acaba de crear una galleta {} y {}".format(sabor, forma))

    def chocolatear(self):
        self.chocolate = True

    def describir(self):
        if self.chocolate:
            print("soy una galleta {}, {} y con chocolate".format(self.sabor, self.forma))
        else:
            print("soy una galleta {}, {} y sin chocolate".format(self.sabor, self.forma))
```

```python
g = galleta("amarga", "cuadrada")
g.describir()
```

```python
g.chocolatear()
g.describir()
```
